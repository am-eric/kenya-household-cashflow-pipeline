net cash flow (week) = income inflows − income outflows − opex outflows

```python
df["trx_value_kes"] = pd.to_numeric(df["trx_value_kes"], errors="coerce")
df["trx_dt"] = pd.to_datetime(df["trx_date"], format="%d%b%Y", errors="coerce")

flows = df[df["trx_class_code"].isin(["income", "opex"])].copy()
flows["signed"] = flows["trx_value_kes"].where(flows["trx_direction"] == "Inflow",
                                               -flows["trx_value_kes"])

h = flows[flows["hh_ids"] == "KMOMM06"].sort_values("trx_dt")
print(h[["trx_dt","trx_class_code","trx_type_desc","trx_direction","trx_value_kes"]].head(40))

weekly = (h.set_index("trx_dt")["signed"].resample("W").sum())
print(weekly.describe())
print(weekly.head(20))
```

```python
key = ["hh_ids", "trx_date", "trx_class_code", "trx_type_desc",
       "trx_direction", "trx_value_kes", "account_ids"]

df["dup_n"] = df.groupby(key, dropna=False)["hh_ids"].transform("size")
df["is_dup_key"] = df["dup_n"] > 1
df["is_cleaning"] = df["interview_designation"] == "06=Cleaning interview"

print(df.groupby("is_cleaning")["is_dup_key"].mean())
```
